```python
# =========================================
# SIMULADOR DE ESTADO DE CUENTA CON "IA"
# Uso de:
# - if
# - elif
# - else
# - while
#
# Restricciones cumplidas:
# ❌ Sin programación orientada a objetos
# ❌ Sin funciones
# ❌ Sin ciclos for
# ❌ Sin try
# =========================================

saldo = 15000
deuda_tarjeta = 4200
prestamo = 10000
ingresos = 18000
gastos = 12000

opcion = 0

print("===================================")
print("  SISTEMA INTELIGENTE FINANCIERO")
print("===================================")

nombre = input("Ingresa tu nombre: ")

print("\nHola", nombre)
print("La IA analizará tu estado de cuenta.\n")

while opcion != 6:

    print("\n========= MENÚ =========")
    print("1. Ver saldo")
    print("2. Depositar dinero")
    print("3. Retirar dinero")
    print("4. Ver análisis inteligente")
    print("5. Simular gasto")
    print("6. Salir")
    print("========================")

    opcion = int(input("Selecciona una opción: "))

    # OPCION 1
    if opcion == 1:

        print("\n----- ESTADO DE CUENTA -----")
        print("Saldo actual: $", saldo)
        print("Deuda tarjeta: $", deuda_tarjeta)
        print("Préstamo pendiente: $", prestamo)

        if saldo > 20000:
            print("IA: Tus finanzas son EXCELENTES.")
        elif saldo >= 10000:
            print("IA: Tus finanzas son ESTABLES.")
        else:
            print("IA: Debes cuidar tus gastos.")

    # OPCION 2
    elif opcion == 2:

        deposito = float(input("¿Cuánto deseas depositar?: $"))

        if deposito > 0:
            saldo = saldo + deposito
            print("Depósito realizado.")
            print("Nuevo saldo: $", saldo)

            if deposito >= 5000:
                print("IA: Excelente ahorro.")
            else:
                print("IA: Todo ahorro ayuda.")
        else:
            print("Cantidad inválida.")

    # OPCION 3
    elif opcion == 3:

        retiro = float(input("¿Cuánto deseas retirar?: $"))

        if retiro <= saldo:

            saldo = saldo - retiro

            print("Retiro realizado.")
            print("Nuevo saldo: $", saldo)

            if retiro > 7000:
                print("IA: Detecto un gasto grande.")
            elif retiro > 3000:
                print("IA: Gasto moderado registrado.")
            else:
                print("IA: Gasto pequeño registrado.")

        else:
            print("Fondos insuficientes.")
            print("IA: No puedes retirar más de lo disponible.")

    # OPCION 4
    elif opcion == 4:

        print("\n===== ANÁLISIS INTELIGENTE =====")

        ahorro = ingresos - gastos

        print("Ingresos mensuales: $", ingresos)
        print("Gastos mensuales: $", gastos)
        print("Capacidad de ahorro: $", ahorro)

        if ahorro > 8000:
            print("IA: Excelente administración financiera.")
        elif ahorro >= 3000:
            print("IA: Tus finanzas son saludables.")
        elif ahorro >= 0:
            print("IA: Debes reducir algunos gastos.")
        else:
            print("IA: Estás gastando más de lo que ganas.")

        if deuda_tarjeta > 5000:
            print("IA: Riesgo alto por deuda en tarjeta.")
        elif deuda_tarjeta > 2000:
            print("IA: Mantén controlada tu deuda.")
        else:
            print("IA: Nivel de deuda bajo.")

        if saldo < 5000:
            print("IA: Tu saldo es bajo.")
        else:
            print("IA: Tienes un saldo aceptable.")

    # OPCION 5
    elif opcion == 5:

        gasto = float(input("Ingresa un gasto inesperado: $"))

        saldo = saldo - gasto

        print("Gasto registrado.")
        print("Saldo restante: $", saldo)

        if saldo < 0:
            print("IA: ALERTA FINANCIERA.")
            print("IA: Tu cuenta está en números rojos.")
        elif saldo < 3000:
            print("IA: Saldo crítico.")
        elif saldo < 10000:
            print("IA: Debes controlar tus gastos.")
        else:
            print("IA: Tu economía sigue estable.")

    # OPCION 6
    elif opcion == 6:

        print("\nGracias por utilizar el sistema.")
        print("IA: Hasta pronto", nombre)

    # OPCION INVALIDA
    else:
        print("Opción inválida.")
```
